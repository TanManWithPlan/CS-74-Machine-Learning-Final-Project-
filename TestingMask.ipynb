This is a very basic experiment to demonstrate that the base BERT model, intially trained on filling in a missing word mask, is not capable of selecting an emotional label.

```python
from transformers import pipeline
```

```python
unmasker = pipeline('fill-mask', model='bert-base-uncased')

text = "Select an emotion from the follow list: 'admiration', 'amusement', 'anger', 'annoyance', 'approval', 'caring', 'confusion', 'curiosity', 'desire', 'disappointment', 'disapproval', 'disgust', 'embarrassment', 'excitement', 'fear', 'gratitude', 'grief', 'joy', 'love', 'nervousness', 'optimism', 'pride', 'realization', 'relief', 'remorse', 'sadness', 'surprise', 'neutral']. The follow text most demonstrates [MASK] emotion: My favourite food is anything I didn't have to cook myself."

unmasker(text)
```
